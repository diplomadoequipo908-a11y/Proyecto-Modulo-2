Generamos un rango de fechas para recorrer publicaciones en automatico.

Definimos la palabra asignada para buscar coincidencias dentro del campo "explanation" de cada publicación APOD

Se hizo un ciclo for para recorrer los años definidos previamente y realizar una solicitud a la API APOD de la NASA para cada año completo, esto con el fin de hacerlo por año y no por día fue reducir el número de solicitudes realizadas ya que la API puede semtirse amenazada.

```python
# Recorremos cada año para evitar miles de solicitudes diarias
for anio in anios:

    # Muestra el avance del proceso
    print(
        "Procesando:",
        anio
    )

    # Definimos fecha inicial y final del año
    inicio=f"{anio}-01-01"

    fin=f"{anio}-12-31"

    # Construimos la URL con API key y rango de fechas
    url=f"https://api.nasa.gov/planetary/apod?api_key={api_key}&start_date={inicio}&end_date={fin}"

    # Enviamos solicitud a la API
    response=requests.get(
        url
    )

    # Verificamos que la solicitud fue exitosa
    if response.status_code==200:

        # Convertimos la respuesta JSON a Python
        data=response.json()

        # Recorremos cada publicación APOD
        for registro in data:

            # Checamos que exista explicación
            if "explanation" in registro:

                # Convertimos texto a minúsculas
                texto=registro[
                    "explanation"
                ].lower()

                # Buscamos la palabra asignada
                if palabra in texto:

                    # Extraemos variables
                    apod_info={

                        "date":
                        registro.get(
                            "date"
                        ),

                        "title":
                        registro.get(
                            "title"
                        ),

                        "media_type":
                        registro.get(
                            "media_type"
                        ),

                        "copyright":
                        registro.get(
                            "copyright"
                        ),

                        "url":
                        registro.get(
                            "url"
                        )

                    }

                    # Convertimos el diccionario en una fila
                    mi_df=pd.DataFrame(
                        [apod_info]
                    )

                    # Agregamos resultados al DataFrame final
                    df_apod=pd.concat(
                        [df_apod,mi_df],
                        ignore_index=True
                    )

    # Pausa para evitar muchas solicitudes
    time.sleep(.3)
```
